# CPRI Prompt Testing Notebook
This notebook loads JSON inputs and prepares the underwriting prompt.

```python
import json
import os

print('Current Directory:', os.getcwd())
print('Files:', os.listdir())
```

```python
# Update this path if needed
BASE_PATH = r"C:\\REPLACE_WITH_YOUR_PATH\\data\\"

def read_json(file_name):
    path = os.path.join(BASE_PATH, file_name)
    print('Reading:', path)
    
    with open(path, 'r', encoding='utf-8-sig') as f:
        content = f.read()
        print('Length:', len(content))
        print('Preview:', repr(content[:100]))
        
        data = json.loads(content)
        print('Loaded successfully')
        return data
```

```python
# Load files
traderisk = read_json('traderisk.json')
assetdescription = read_json('assetdescription.json')
scoring = read_json('scoring_stub.json')
```

```python
# Merge all inputs
payload = {}
payload.update(traderisk)
payload.update(assetdescription)
payload.update(scoring)

print('Final Payload Keys:', list(payload.keys())[:20])
```

```python
# Example: Build prompt (simplified)
prompt = f"Deal Type: {payload.get('deal_type')}\nCountry: {payload.get('country_of_risk')}"
print(prompt)
```
